```python
text_columns = [
    "product_name",
    "brand",
    "categories",
    "short_description",
    "full_description"
]

for column in text_columns:
    df[column] = df[column].str.replace(r"\s+", " ", regex=True).str.strip()
```

```python
print("Current prices:")
print(df["current_price"].value_counts().head(20))

print("\nOriginal prices:")
print(df["original_price"].dropna().value_counts().head(20))
```

```python
def clean_price(price):
    if pd.isna(price):
        return None

    price = price.replace("DA", "").strip()
    price = price.replace(".", "")
    price = price.replace(",", ".")

    return float(price)


df["current_price"] = df["current_price"].apply(clean_price)
df["original_price"] = df["original_price"].apply(clean_price)
```

```python
# Create a working copy
cleaned_df = df.copy()

# Missing brand
cleaned_df["brand"] = cleaned_df["brand"].fillna("Unknown")

# Missing text descriptions
cleaned_df["short_description"] = cleaned_df["short_description"].fillna("")
cleaned_df["full_description"] = cleaned_df["full_description"].fillna("")

# Keep missing prices as NaN (original_price and current_price are not imputed)

print("Missing values after cleaning:\n")
print(cleaned_df.isna().sum())
```

```python
# Zero prices -> missing
df.loc[df["current_price"] == 0, "current_price"] = pd.NA
#3 products have a current price of 0 
```

```python
matches_name = (
    df["full_description"].str.strip().str.lower()
    == df["product_name"].str.strip().str.lower()
)

print("Descriptions that repeat the name:", matches_name.sum())   # expect 36

df.loc[matches_name, "full_description"] = pd.NA
```
